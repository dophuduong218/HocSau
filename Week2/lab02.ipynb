## Bài 5. Reshape và Flatten

## Bài 6. Phép toán và nhân ma trận

```python
import torch

torch.manual_seed(42)
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\na * b (nhân từng phần tử):")
print(a * b)

print("\na @ b.T (nhân ma trận):")
print(a @ b.T)
print("Shape của a @ b.T:", (a @ b.T).shape)
```

## Bài 7. Các hàm thống kê trên Tensor

```python
import torch

torch.manual_seed(42)
a = torch.randn(3, 4)

print("a:")
print(a)

print("\nMean theo cột (dim=0):")
print(torch.mean(a, dim=0))

print("\nStandard deviation theo cột (dim=0):")
print(torch.std(a, dim=0))

print("\nCumulative sum theo dim=0:")
print(torch.cumsum(a, dim=0))

print("\nMean theo hàng (dim=1):")
print(torch.mean(a, dim=1))

print("\nStandard deviation theo hàng (dim=1):")
print(torch.std(a, dim=1))

print("\nCumulative sum theo dim=1:")
print(torch.cumsum(a, dim=1))
```

## Bài 8. Autograd cơ bản

```python
import torch

# y = x^2
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("Trường hợp y = x^2")
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# y = 3*x^2 + 2*x + 1
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()

print("\nTrường hợp y = 3*x^2 + 2*x + 1")
print("x =", x2)
print("y =", y2)
print("x.grad =", x2.grad)
```

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
print("Final MSE:", mse.item())
```

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import torch

torch.manual_seed(42)

A = torch.tensor(torch.rand(()).item(), requires_grad=True)
B = torch.tensor(torch.rand(()).item(), requires_grad=True)
C = torch.tensor(torch.rand(()).item(), requires_grad=True)
D = torch.tensor(torch.rand(()).item(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

print("\nKiểm tra:")
print("A + B =", (A+B).item())
print("C - D =", (C-D).item())
print("A + C =", (A+C).item())
print("B - D =", (B-D).item())
print("Tổng sai số bình phương =", sqerr.item())
```

## Bài 11. Bài tập áp dụng

```python
import numpy as np
import torch

# 1. Tensor X có 24 phần tử và reshape
X = torch.arange(1, 25)

print("X ban đầu:", X)
print("4 x 6:")
print(X.reshape(4, 6))
print("2 x 12:")
print(X.reshape(2, 12))
print("2 x 3 x 4:")
print(X.reshape(2, 3, 4))

# 2. Hai Tensor A, B kích thước 3 x 4
torch.manual_seed(42)
A = torch.randn(3, 4)
B = torch.randn(3, 4)

print("\nA:")
print(A)
print("B:")
print(B)
print("A + B:")
print(A + B)
print("A * B:")
print(A * B)
print("A / B:")
print(A / B)
print("Trung bình theo cột:")
print(torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột:")
print(torch.std(A, dim=0))

# 3. y = 2x^2 + 5x + 3 tại x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 2*x**2 + 5*x + 3
y.backward()

print("\ny = 2x^2 + 5x + 3")
print("y tại x=2:", y.item())
print("dy/dx bằng autograd:", x.grad.item())
print("dy/dx thủ công = 4x + 5 =", 4*2 + 5)

# 4. Huấn luyện đa thức y = 2x^2 - 3x + 5
np.random.seed(42)
torch.manual_seed(42)

N = 20
X_np = np.random.randn(N, 1) * 5
Y_np = 2 * X_np**2 - 3 * X_np + 5

XX = np.hstack([X_np**2, X_np, np.ones_like(X_np)])

w = torch.randn(3, 1, requires_grad=True)
x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y_np, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x_train @ w
    mse = torch.mean((y_train - y_pred)**2)
    mse.backward()
    optimizer.step()

print("\nHệ số tìm được cho y = 2x^2 - 3x + 5:")
print(w)
print("Final MSE:", mse.item())
```
